### EJERCICIO 1-Obtener la inversa de la matriz.

```python
def Jordan_gaussiana(A: np.ndarray) -> np.ndarray:
    """
    Aplica el método de Gauss–Jordan normalizado y devuelve
    la matriz en forma reducida por filas (RREF).
    """

    if not isinstance(A, np.ndarray):
        A = np.array(A, dtype=float)

    n, m = A.shape
    assert m >= n, "La matriz debe ser aumentada."

    for i in range(n):
        # --- buscar pivote
        p = None
        for pi in range(i, n):
            if A[pi, i] != 0:
                p = pi
                break

        if p is None:
            raise ValueError("No existe solución única.")

        # --- intercambio de filas
        if p != i:
            A[[i, p]] = A[[p, i]]

        # --- normalizar pivote
        pivote = A[i, i]
        A[i, :] = A[i, :] / pivote

        # --- eliminar arriba y abajo
        for j in range(n):
            if j != i:
                factor = A[j, i]
                A[j, :] = A[j, :] - factor * A[i, :]

    return A
```

```python
def inversa_por_jordan_normalizado(A: np.ndarray) -> np.ndarray:
    n = A.shape[0]
    AI = np.hstack((A, np.eye(n))) ## Crear matriz aumentada [A | I] Horizontalmente
    R = Jordan_gaussiana(AI)

    A_inv = R[:, n:]
    return A_inv
```

### Matriz 1

```python
A01 = np.array([
    [1, 3, 4],
    [2, 1, 3],
    [4, 2, 1]
])
```

```python
A01_inv=inversa_por_jordan_normalizado(A01) 
print("La inversa de A01 es:\n", A01_inv)
```
